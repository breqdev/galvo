```python
from svgpathtools import svg2paths, wsvg

# paths, attributes = svg2paths('breqcube.svg')
paths, attributes = svg2paths('avalogo.svg')

points = []

for path in paths:
    # x = [line.start.real for line in path]
    # x.append(path[-1].end.real)
    # y = [line.start.imag for line in path]
    # y.append(path[-1].end.imag)

    first_point = True

    for line in path:
        for i in range(int(line.length())):
            point = line.point(i / int(line.length()))
            points.append((point.real, point.imag, not first_point))
            first_point = False

    last = path[-1].point(1)
    points.append((last.real, last.imag, True))

for (x, y, pen) in points:
    # print(f"{'+' if pen else '-'} {int(x)} {int(y)}")
    print(f"{'+' if pen else '-'} {int((x - 87.5) * 10 + 128)} {int((y - 60) * 10 + 128)}")
```
